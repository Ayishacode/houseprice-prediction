```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df1=pd.read_csv("Bengaluru.csv")
df1.head()
```

```python
df1.shape
```

```python
df2=df1.drop(['area_type','society','balcony','availability'],axis=1)
df2.head()
```

```python
df2.isnull().sum()
```

```python
df3=df2.dropna()
df3.isnull().sum()
```

```python
df3['size'].unique()
```

```python
df3['size']=df3['size'].apply(lambda x: int(x.split(' ')[0]))
df3.head()
```

```python
df3['total_sqft'].unique()
```

```python
def is_float(x):
    try:
        float(x)
    except:
        return False
    return True
df3[~df3['total_sqft'].apply(is_float)].head(10)
```

```python
def convert(x):
    tokens=x.split('-')
    if len(tokens)==2:
        return ((float(tokens[0])+float(tokens[1]))/2)
    try:
        return float(x)
    except:
        return None
df4=df3.copy()        
df4['total_sqft']=df4['total_sqft'].apply(convert)
df4.head()
```

```python
df4.loc[30]
```

```python
df5=df4.copy()
df5['price_per_sqft']=df5['price']*100000/df5['total_sqft']
df5.head()
```

```python
df5.location = df5.location.apply(lambda x: x.strip())
location_stats = df5['location'].value_counts(ascending=False)
location_stats
```

```python
len(location_stats[location_stats>10])
```

```python
len(df5.location.unique())
```

```python
location_stats_less_than_10 = location_stats[location_stats<=10]
location_stats_less_than_10
```

```python
df5.location = df5.location.apply(lambda x: 'other' if x in location_stats_less_than_10  else x)
len(df5.location.unique())
```

```python
df5.head()
```

```python
#removing outliers
df5.shape
```

```python
df6 = df5[df5['total_sqft'] / df5['size'] >= 300]
df6.shape
```

```python
#Because real estate prices vary drastically by location, filter out price_per_sqft values beyond 1 standard deviation ($\sigma$) from the location's mean ($\mu$).
def remove_pps_outliers(df):
    df_out = pd.DataFrame()
    for key, subdf in df.groupby('location'):
        m = np.mean(subdf['price_per_sqft'])
        st = np.std(subdf['price_per_sqft'])
        # Keep points within 1 standard deviation
        reduced_df = subdf[(subdf['price_per_sqft'] > (m - st)) & (subdf['price_per_sqft'] <= (m + st))]
        df_out = pd.concat([df_out, reduced_df], ignore_index=True)
    return df_out

df7 = remove_pps_outliers(df6)
df7.shape
```

```python
#Filter out properties where a larger BHK costs less per sq ft than a smaller BHK in the same location (e.g., a 3 BHK priced lower per sq ft than the mean of 2 BHKs in that location).
def remove_bhk_outliers(df):
    exclude_indices = np.array([])
    for location, location_df in df.groupby('location'):
        bhk_stats = {}
        for bhk, bhk_df in location_df.groupby('size'):
            bhk_stats[bhk] = {
                'mean': np.mean(bhk_df.price_per_sqft),
                'std': np.std(bhk_df.price_per_sqft),
                'count': bhk_df.shape[0]
            }
        for bhk, bhk_df in location_df.groupby('size'):
            stats = bhk_stats.get(bhk-1)
            if stats and stats['count']>5:
                exclude_indices = np.append(exclude_indices, bhk_df[bhk_df.price_per_sqft<(stats['mean'])].index.values)
    return df.drop(exclude_indices,axis='index')
df8 = remove_bhk_outliers(df7)
df8.shape
```

```python
# Keep properties where bathrooms don't exceed bedrooms + 2
df8 = df8[df8['bath'] < df8['size'] + 2]
df8.shape
```

```python
df8['bhk']=df8['size']
df8=df8.drop('size', axis=1)
df8.head()
```

```python
#model training
#one hot encoding
dummies=pd.get_dummies(df8.location,dtype=int)
dummies.head()
```

```python
#dropping the col other to avoid dummy variable trap
dummies = dummies.drop('other', axis=1)
#concat the data to original 
df8 = pd.concat([df8, dummies], axis=1)
df8 = df8.drop('location', axis=1)
df8.head()
```

```python
x=df8.drop(['price','price_per_sqft'],axis=1)
x
```

```python
y=df8.price
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
score = model.score(x_test, y_test)
score
```

```python
#CROSS VALIDATION 
#Instead of splitting your data into a single train and test set (e.g., $80/20$), cross-validation splits the data multiple times, trains the model on different subsets, and averages the performance scores.
from sklearn.model_selection import ShuffleSplit, cross_val_score

# Randomize split across 5 folds
cv = ShuffleSplit(n_splits=5, test_size=0.2, random_state=0)
cv_scores = cross_val_score(LinearRegression(), x, y, cv=cv)

print("5-Fold Scores:", cv_scores)
print(f"Mean Score: {cv_scores.mean():.4f}")
```

```python
#python fn to link the model and user input
def predict_price(location, sqft, bath, bhk):
    loc_index = np.where(x.columns == location)[0][0] if location in x.columns else -1
    x_input = np.zeros(len(x.columns))
    x_input[0] = sqft
    x_input[1] = bath
    x_input[2] = bhk
    
    if loc_index >= 0:
        x_input[loc_index] = 1
    x_df = pd.DataFrame([x_input], columns=x.columns)
    
    return model.predict(x_df)[0]
```

```python
predict_price('1st Phase JP Nagar',1000, 2, 2) 
```
